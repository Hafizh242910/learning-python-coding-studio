```python
# List = Menyimpan kumpulan item berurutan
# membuat lis -> []
# mutable -> bisa kita ubah

nama_siswa = ["clairine", "fadlan", "gio", "anton", "hafizh"]
print("List nama siswa:", nama_siswa)
nama_siswa_kosong = [] # List kosong

# menggunakan method untuk menambah dan menghapus data di dalam list
# method menambah -> append(), insert()

# method append()
nama_siswa.append("budi") # ini untuk menambah  data budi
print("nama siswa baru ditambahkan:", nama_siswa)

# method insert () / menambah data di posisi tertentu
nama_siswa.insert(1,"andi") # insert(posisi, nama-value)
print("nama siswa andi ditambah di posisi 2:", nama_siswa)

# method remove() dan pop()
# method remove () -> kita harus mengetahui value yang ingin di hapus
nama_siswa.remove("fadlan") # menghapus data
print("Update list:", nama_siswa)

# method pop() -> menghapus berdasarkan index/posisi
nama_siswa.pop(5)
print("Update list:", nama_siswa)

# List indexing -> mengakses elemen tertentu di dalam list
# nama_var[posisi_index]
print(nama_siswa[2]) # ini akan menampilkan posisi 3 dari data list

# List slicing -> memotong list = mengakses beberapa value dalam list
print(nama_siswa[0:3]) # [Star:stop]
print(nama_siswa[1:]) # start:
print(nama_siswa[:2]) # :end
print(nama_siswa[-1]) # :end

# melihat panjang list
print("panjang list:", len(nama_siswa))

```

```python
# Set -> himpunan -> {}
# tidak bisa menerima duplicate elemen
# tidak ada urutan

himpunan_nama = {"gio", "fadlan", "clairine"}
print("set awal:", himpunan_nama)
# menambah data ke dalam set -> add()
himpunan_nama.add("anton")
print("set update:", himpunan_nama)

# menghapus data dalam set
# method remove()
himpunan_nama.remove("fadlan") # error baris kode dibawah tdk akan dijalankan
print("set update setelah di remove:", himpunan_nama)

# method discard
himpunan_nama.discard("ANTON")
print("set update setelah di remove:", himpunan_nama)

# Gabungan / join, selisih, irisan/intersection
himp_siswa_a = {"clairine", "fadlan", "gio"}
himp_siswa_b = {"fadlan", "hafizh", "anton"}

# gabungan/join -> |
print("gabungan kedua himp:", himp_siswa_a | himp_siswa_b)

# selisih -> -
print("selisih himp a dan himp b:", himp_siswa_a - himp_siswa_b)
print("selisih himp b dan himp a:", himp_siswa_b - himp_siswa_a)

# irisan ->&
print("irisan himpunan:", himp_siswa_a & himp_siswa_b)
```

```python
# Dictionary -> kamus {key:value}

kontak_telpon = { # key:value
    "clairine": 998,
    "fadlan": 997,
    "gio": 996
}
print("dictionary awal:", kontak_telpon)
# menambahkan data ke dalam dictionary -> indexing
kontak_telpon["hafizh"] = 888
print("dictionary update:", kontak_telpon)

# merubah data dengan indexing
kontak_telpon["gio"] = 889
print("dictionary update:", kontak_telpon)

# menghapus data dalam dict
# pop()
kontak_telpon.pop("gio")
print("dictionary update:", kontak_telpon)

# del
del kontak_telpon["fadlan"]
print("dictionary update:", kontak_telpon)

# keys(), values(), items()
# keys() -> mengambil semnua key()
print(kontak_telpon.keys())

# values() -> mengambil semua value
print(kontak_telpon.values())

# item () -> mengambil semua key dan value
print(kontak_telpon.items())
```

```python
# Chatbot Sederhana
import random

# Respon pertanyaan
# key: value (pertanyaan:jawaban)
respons = {
    "hai": ["Halo apa kabar?", "Halo", "Selamat Datang"],
    "nama kamu siapa": ["Saya Chatbot Gacor", "Saya Pembantu Anda"],
    "bye": ["Selamat Tinggal", "Sampai Jumpa", "Bye"]
}

def chatbot():
    # mintak input (user input)
    # cek kondisi (inputan = data di respons)

    # perulangan unttuk mengecek kondisi
    while True: #ini kondisi benar
        user_input = input("User: ").lower()

        # cek user input apakah ada di respons
        if user_input in respons:
            print("Bot: ", random.choice(respons[user_input]))
        elif user_input == "keluar":
            print("Bot: Sampai Jumpa")
            break
        else:
            print("Bot: Maaf Saya Tidak Mengerti")

# panggil function
chatbot()
        
```
