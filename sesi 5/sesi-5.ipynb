```python
"""
sintax kelas di python

class NamaKelas:
    isi dari kelas
    # mengisi atribut/data dari kelas
    def__init__(self, atribut1, atribut2, dst)
     # self untuk mengarah ke kelas itu sendiri
      self.atribut1 = atribut1
      dst

    # method di dalam kelas  
    def nama_method():
      isi method

sintax membuat object:

nama_object = NamaKelas(isiatribut1 dst)
nama_object.nama_method atau nama_atribut
"""

# nama kelas
class Mobil:
    def __init__(self, merk, warna, tahun):
      self.merk = merk # atribute/data merk
      self.warna = warna # atribute/data warna
      self.tahun = tahun # atribute/data tahun

    def jalan(self):
      print(f"mobil {self.merk} sedang berjalan")

    def berhenti(self):
      print(f"mobil {self.merk} berhenti")

    def klakson(self):
      print(f"mobil {self.warna} klakson")

# membuat objek dari kelas
mobil1 = Mobil("Toyota", "Hijau", 2020) # set atribut 
mobil2 = Mobil("Mitsubhisi", "Putih", 2020)
mobil3 = Mobil("Honda", "Hitam", 2025)

# memanggil method dari kelas menggunakan titik dan objek
mobil1.jalan()
mobil2.jalan()
mobil3.jalan()

# memanggil method dari kelas menggunakan titik dan objek
mobil1.berhenti()
mobil2.berhenti()
mobil3.berhenti()

# memanggil method dari kelas menggunakan titik dan objek
mobil1.klakson()
mobil2.klakson()
mobil3.klakson()

# kita panggil atribut/data
print(mobil1.warna)
print(mobil1.tahun)
mobil1.tahun = 2025
print(mobil1.tahun)
#atribut itu engga pakai ()
```

```python
# Encapsulation = menyimbunyikan data / atribut agar tidak bisa diakses
# biasanya digunakan untuk melindungi data

class AkunBank:
    def __init__(self, nama, saldo):
      self.nama = nama # public atribute
      self.__saldo = saldo # private saldo atribut
      # perhatikan penggunaan __

    # setter dan getter
    def setSaldo(self, saldo):
      self.__saldo = saldo

    def getSaldo(self):
      print(f"Saldo kamu {self.__saldo}")
    
# membuat objek/akun
akun1 = AkunBank("Hafizh", 7000)
print(akun1.nama)
akun1.nama = "Anton"
print(akun1.nama)

akun1.getSaldo()
```

```python
# Inheritance (pewarisan)
#-> sebuah kelas bisa mewarisi atribut dan method dari class lain.
# mirip seperti anak mewarisi sifat dari orang tua

# parent kelas/super class
class Hewan:
    # atributnya
    def __init__(self, nama):
      self.nama = nama

    def suara(self):
      print("Hewan Bersuara")

    def jumlah_kaki(self):
      print("Hewan berkaki 4")

    def melahirkan(self):
      print("Hewan Beranak")

# child kelas
class Kucing(Hewan): # Kelas kucing mewarisi dari kelas hewan
  def suara(self):
    print("Meong-meong")

# child kelas
class Anjing(Hewan): # Kelas anjing mewarisi dari kelas hewan
  def suara(self):
    print("Guk-guk")

# buat object
mimi = Kucing("mimi")
leki = Anjing("leki")
mimi.suara()
leki.suara()
leki.jumlah_kaki()
mimi.melahirkan()
```

```python
# Polimorfisme
# objek berbeda bisa memimiliki method dengan nama sama,
# tapi implementasinya berbeda.

class Person:
    def __init__(self, nama):
      self.nama = nama

    def greet(self):
      print("Halo teman-teman")

class English(Person):
   def greet(self): # ovveride
     print("Halo friends")

class German(Person):
   def greet(self):
     print("tschuss preunds")

# buat object
person1 = English("Hafizh")
person2 = German("Gio")

person1.greet() # nama method sama
person2.greet() # tapi implementasi beda
```
