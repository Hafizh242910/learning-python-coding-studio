```python
# control flow if elif else
'''
sintax
if kondisi = true:
  blok kode yang akan dieksekusi
elif kondisi = true:
  blok kode yang akan dieksekusi
elif kondisi = true:
  blok kode yang akan dieksekusi
else: #apabila if dan elif false
  blok kode yang akan di eksekusi
'''
nilai_siswa = 88

if nilai_siswa >= 90: # comparison
    print("A")
elif nilai_siswa >= 80:
    print("A")
elif nilai_siswa >= 70:
    print("C")
else:
    print("D")
```

```python
1 # operator comparison (hasil true/ false)
# <, >, <=, >=, ==, !=
a = 10
b = 5
print(a > b) # True
print(a >= b) # True
print(a < b) # False
print(a <= b) # False
print(a == b) # False
print(a != b) # True
```
